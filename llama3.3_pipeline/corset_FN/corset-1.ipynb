```python
# Import required libraries
import pandas as pd
from google.colab import drive

# Step 1: Mount Google Drive to access files
drive.mount('/content/drive')

# Define your file paths
# (Note: If your output.csv is inside a specific folder in your Drive, update the path accordingly)
input_file_path = '/content/drive/MyDrive/output.csv'
output_file_path = '/content/drive/MyDrive/coreset-1.csv'

# Step 2: Load the dataset
df = pd.read_csv(input_file_path)

# Step 3: Filter instances where the 'model_answer' column has "No" in it
# Using str.contains to catch variations (e.g., "No", "No, it isn't"), na=False ignores missing values
filtered_df = df[df['model_answer'].astype(str).str.contains('No', na=False)]

# Step 4: Select a random coreset of 600 instances
# We check if there are at least 600 rows to avoid an error.
# 'random_state=42' ensures you get the exact same random 600 rows if you run this cell again.
if len(filtered_df) >= 600:
    coreset_df = filtered_df.sample(n=600, random_state=42)
    print("Successfully selected 600 random instances.")
else:
    print(f"Warning: Only {len(filtered_df)} instances match the criteria. Selecting all available instances.")
    coreset_df = filtered_df.sample(frac=1, random_state=42) # Shuffles the remaining rows

# Step 5: Save the resulting dataframe to a new CSV file
coreset_df.to_csv(output_file_path, index=False)
print(f"Coreset saved successfully to: {output_file_path}")
```
